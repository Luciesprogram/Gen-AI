```python
from langchain_core.messages import trim_messages, SystemMessage, AIMessage
```

```python
trimmer = trim_messages(
    max_tokens= 100,
    strategy="last", #last 40 tokens, first, random
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
```

```python
from operator import itemgetter

from langchain_core.runnables import RunnablePassthrough

def filter_messages(messages):
    # Example: Only keep messages that have more than 5 characters
    # and exclude any specific metadata you don't like.
    return [msg for msg in messages if len(msg.content) > 5]

chain = (
    RunnablePassthrough.assign(messages=itemgetter("messages") | RunnableLambda(filter_messages) | trimmer)
    | prompt
    | model

)
```
